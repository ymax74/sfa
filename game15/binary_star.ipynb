```python
import matplotlib.pyplot as plt
%matplotlib widget

from math import *

import rebound
```

```python
sim = rebound.Simulation()
sim.units = ('yr', 'AU', 'Msun') 
sim.add(m=0.7, x=0.0,y=0.0,z=0.0, vx = 0, vy = 0.0, vz = 0.)
sim.add(m=0.3, x=1.0,y=0.0,z=0.0, vx = 0, vy = 7.0, vz = 0.)
sim.N_active = sim.N
sim.add(m=0.1, x=1.2,y=0.45,z=0.0, vx = 3.0, vy = 6.0, vz = 0.)
sim.status()
# op = rebound.OrbitPlot(sim, orbit_style=None)
op = rebound.OrbitPlot(sim, orbit_style=None, xlim=[-2, 2], ylim=[-2, 2])

print(sim.N_active)

# o = sim.particles[1].calculate_orbit(primary=sim.particles[0])
o = sim.particles[1].orbit(primary=sim.particles[0])
# print(o.P)
```

```python
fig = op.fig
for i in range(1000):
    op.sim.integrate(sim.t+0.001)
    op.sim.move_to_hel()
    op.update()       # update data
    fig.canvas.draw() # redraw figure
```

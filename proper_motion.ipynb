## Учет собственных движений звезд

```python
from math import *
import numpy as np
```

```python
def angular_distance(ra,dec,RA,DEC):
    return acos(sin(dec)*sin(DEC)+cos(dec)*cos(DEC)*cos(ra-RA))

def tangential_coordinates(ra,dec,RA,DEC):
    ksi = cos(dec)*sin(ra-RA)/(sin(dec)*sin(DEC)+cos(dec)*cos(DEC)*cos(ra-RA))
    eta = (sin(dec)*cos(DEC)-cos(dec)*sin(DEC)*cos(ra-RA))/(sin(dec)*sin(DEC)+cos(dec)*cos(DEC)*cos(ra-RA))
    return ksi,eta

def equatorial_coordinates(ksi,eta,RA,Dec):
    x,y,z = np.dot(np.array([[-sin(RA),-cos(RA) * sin(Dec),cos(RA) * cos(Dec)],
                             [cos(RA),-sin(RA) * sin(Dec),sin(RA) * cos(Dec)],
                             [0,cos(Dec),sin(Dec)]]),
                   np.array([ksi,eta,1]))/sqrt(1+ksi*ksi+eta*eta)
    ra = atan2(y,x)
    dec = atan2(z,sqrt(x*x+y*y))
    if(ra<0):
        ra+=2*pi
    return ra,dec

def proper_motion(ra,dec,pmra,pmdec, t, epoch):
    ksi,eta = pmra*(t - epoch)/3.6e6,pmdec*(t - epoch)/3.6e6
    ra_t,dec_t = equatorial_coordinates(radians(ksi),radians(eta),radians(ra),radians(dec))
    return degrees(ra_t),degrees(dec_t)
    
```

```python
pmra,pmdec = 7,10 #mas/yr
t = 2023.5
epoch = 2000.0
```

```python
Np=2000
dec = np.linspace(-90,90,Np)
ra = 0
ra1,dec1 = np.zeros(np.size(dec)),np.zeros(np.size(dec))
ra2,dec2 = np.zeros(np.size(dec)),np.zeros(np.size(dec))
for count,DEC in enumerate(dec):
    ra1[count] = ra+pmra*(t - epoch)/3.6e6/cos(radians(DEC))
    dec1[count] = DEC+pmdec*(t - epoch)/3.6e6
    ra2[count],dec2[count]=proper_motion(ra,DEC,pmra,pmdec, t, epoch)
```

```python
import matplotlib.pyplot as plt
%matplotlib widget
fig, ax = plt.subplots(figsize=(2,1), dpi=300)
# plt.scatter(dec,(ra2-ra1)*3.6e6, c='green', s=0.5,   edgecolors='none')
plt.plot(dec,(dec2-dec1)*3.6e6, c='green', linewidth=0.3)
ax.xaxis.set_tick_params(labelsize=5)
ax.yaxis.set_tick_params(labelsize=5)
ax.set_xlim(-90,90)
ax.set_ylim(-50,50)
ax.set_xlabel('$\\delta$, (deg)', fontsize=4)
ax.set_ylabel('$\\Delta \\delta$, (mas)', fontsize=4)
plt.tight_layout()
```
